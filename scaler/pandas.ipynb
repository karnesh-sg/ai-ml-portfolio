```python
import numpy as np

def stock_analysis(stocks):

    # Percentage change between consecutive days
    percentage_change = ((stocks[:, 1:] - stocks[:, :-1]) / stocks[:, :-1]) * 100
    percentage_change = np.round(percentage_change, 2)

    # Companies with at least one negative change
    negative_companies = np.where(np.any(percentage_change < 0, axis=1))[0]

    # Days with more than 5% increase
    days_gt_5 = percentage_change > 5

    # Print the output in required format
    print("Percentage Change:")
    print(percentage_change)

    print("\nCompanies with at least one negative change:")
    print(negative_companies)

    print("\nDays with >5% increase:")
    print(days_gt_5)
```
